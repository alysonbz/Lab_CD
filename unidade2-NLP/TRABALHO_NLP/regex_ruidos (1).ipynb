**Etapa 2 — CRIAÇÃO DE PADRÕES REGEX**

Essa célula define um conjunto de expressões regulares (regex) para identificar diferentes tipos de informações dentro de textos com ruídos. Cada "re.compile()" cria um objeto regex pronto para ser usado em buscas e substituições.

rx_email: padrão criado para validar a coluna e_mails, onde temos o início da string, parte local do e-mail. @, que será um símbolo obrigatório, o domínio (ex: gmail, outlook), e como irá terminar, sendo com ponto + TLD (ex: .com, .org, .br).

RX_CPF: padrão criado para validar a coluna cpf, onde detecta CPFs com ou sem pontos e traço, e define (?<!\d) e (?!\d) que garantem que o padrão não está dentro de um número maior.

RX_CNPJ: padrão criado para validar a coluna cpf, onde vai capturar CNPJs, com ou sem formatação (pontos, barras e traço).

RX_CEP: padrão criado para validar a coluna cep, detecta CEPs (ex: 12345-678 ou 12345678).

RX_DATA: padrão criado para validar a coluna data, onde irá identificar datas no formato brasileiro (DD/MM/AAAA), e permite 1 ou 2 dígitos para dia e mês (ex: 1/1/2024 ou 01/01/2024).

RX_FONE: padrão criado para validar a coluna telefone, que detecta números de telefone brasileiros, com ou sem DDD e com ou sem o +55. Permite formatos variados: +55 (11) 98765-4321, 11987654321, 9876-4321 etc.

RX_EMAIL: padrão criado para detectar emails, explicado anteriormente.

RX_URL:  padrão criado para detectar URLs, que encontra URLs que comecem com http:// ou https://.

RX_NOME: padrão criado para detectar nomes, detecta nomes de pessoas, permitindo acentuação. Exige pelo menos duas palavras (ex: “Maria Silva”).

RX_TELEFONE: a mesma coisa do padrão RX_FONE.

```python
rx_email = re.compile(r"""
    ^[A-Za-z0-9._%+-]+      # parte local
    @
    [A-Za-z0-9.-]+          # domínio
    \.[A-Za-z]{2,}$         # TLD
""", re.X)

RX_CPF = re.compile(r"(?<!\d)(?:\d{3}\.?\d{3}\.?\d{3}-?\d{2})(?!\d)")
RX_CNPJ = re.compile(r"(?<!\d)(?:\d{2}\.?\d{3}\.?\d{3}/?\d{4}-?\d{2})(?!\d)")
RX_CEP = re.compile(r"(?<!\d)\d{5}-?\d{3}(?!\d)")
RX_DATA = re.compile(r"(?<!\d)(0?[1-9]|[12]\d|3[01])/(0?[1-9]|1[0-2])/\d{4}(?!\d)")
RX_FONE = re.compile(r"(?x)(?:\+?55\s*)?(?:\(?\d{2}\)?\s*)?(?:9?\d{4})-?\s?\d{4}")
RX_EMAIL = rx_email
RX_URL = re.compile(r"https?://[^\s/$.?#].[^\s]*", re.I)
RX_NOME = re.compile(r"[A-Za-zÀ-ÖØ-öø-ÿ]+(?:\s+[A-Za-zÀ-ÖØ-öø-ÿ]+)+")
RX_TELEFONE = re.compile(
    r"(?:\+?55\s*)?(?:\(?\d{2}\)?\s*)?\d{4,5}-?\d{4}"
)
```

**Etapa 3 — NORMALIZAÇÃO**

Para começarmos a normalização, definos uma função chamada "so_digitos" que recebe uma string s e retorna outra string contendo apenas dígitos (números de 0 a 9), logo após, ela substitui qualquer caractere não numérico (\D) por nada (""). O uso de s or "" garante que, se s for None, ele substitui por uma string vazia antes de aplicar o regex. Essa função é útil para padronizar CPFs, CNPJs, telefones ou CEPs, retirando toda a formatação.

```python
def so_digitos(s: str) -> str:
    return re.sub(r"\D", "", s or "")
```

Fazemos primeiramente a normalização dos nomes. Criamos uma função que limpa e corrige nomes, com o nome de "normaliza_nome". Ela irá retornar None se o texto for inválido, ou o nome normalizado se tudo der certo. Se o valor não for uma string (por exemplo, NaN ou número), retorna None.

Usando a função "re.sub", mantemos apenas letras (inclusive acentuadas) e espaços, remove números, pontuações e símbolos estranhos. Depois, substitui vários espaços por um único espaço e elimina espaços nas bordas. Se não, retorna None.

Dividimos o nomes em palavras.

Depois, fazemos um Loop para corrigimos letras soltas, que irá juntar letras soltas com a próxima palavra (corrige erros como “H elena” = “Helena”), e ignora letras sozinhas no final, mantendo palavras normais. Depois, irá retornar os nomes formatados corretamente, junta as palavras de volta com um espaço e coloca cada palavra com a primeira letra maiúscula (title()).

Por ultimo, aplica a função como uma nova coluna no nosso dataset. Cria uma nova coluna nome_normalizado com os nomes limpos.

```python
def normaliza_nome(texto: str) -> str | None:
    if not isinstance(texto, str):
        return None

    # 1) Mantém só letras (com acento) e espaços
    s = re.sub(r"[^A-Za-zÀ-ÖØ-öø-ÿ\s]", " ", texto)
    s = re.sub(r"\s+", " ", s).strip()
    if not s:
        return None

    tokens = s.split()
    merged = []
    i = 0

    while i < len(tokens):
        t = tokens[i]

        # se for letra única e tiver próxima palavra maior, junta as duas
        if len(t) == 1 and i + 1 < len(tokens) and len(tokens[i+1]) > 1:
            merged.append(t + tokens[i+1])   # "H" + "elena" -> "Helena"
            i += 2
        elif len(t) == 1:
            # letra sozinha no fim ou seguida de outra letra sozinha → ignora
            i += 1
        else:
            merged.append(t)
            i += 1

    if not merged:
        return None

    return " ".join(merged).title()

df["nome_normalizado"] = df["nome"].apply(normaliza_nome)
```

Agora, iremos fazer a normalização e formatação do padrão númerico dos CPFs, onde iremos definir uma função auxiliar para formatar uma string de 11 dígitos em formato de CPF.


Temos um return que insere os pontos e o traço no formato clássico:
Exemplo: "12345678900" → "123.456.789-00"


Fazemos a função principal que limpa e valida o CPF. Nela, usamos a função definida antes (so_digitos) para remover caracteres não numéricos, ficando só com os números.


Fazemos um len que, qualquer CPF válido deve ter exatamente 11 dígitos.
Se não tiver, retorna None e será considerado invalido. Na saída, caso o tamanho esteja correto, retorna o CPF já formatado.


Depois, aplica essa função à coluna "cpf" e cria uma nova coluna "cpf_normalizado" no Dataset.

```python
def formata_cpf(d: str) -> str:
    return f"{d[:3]}.{d[3:6]}.{d[6:9]}-{d[9:]}"

def normaliza_cpf(cpf: str) -> str | None:
    d = so_digitos(cpf)
    if len(d) != 11:
        return None
    return formata_cpf(d)

df["cpf_normalizado"] = df["cpf"].apply(normaliza_cpf)
```

Agora, iremos fazer a normalização e formatação do padrão númerico dos CNPJs, onde iremos definir uma função auxiliar para formatar uma string de 14 dígitos em formato de CNPJ.

Fazemos uma função para decidirmos o formato, que cria uma função para formatar CNPJs, que devem ter 14 dígitos. Exemplo: "12345678000195" → "12.345.678/0001-95"

Depois, fazemos a função principal para normalizar os CNPJs para limpar e validar o CNPJ. Onde definimos d, que remove qualquer caractere que não seja número.

E um len Verifica se o CNPJ tem o número correto de dígitos. Se não tiver, é inválido.


Verificamos se estiver válido, devolve o CNPJ formatado.


Por ultimo, criamos uma nova coluna com os CNPJs normalizados e adicionamos no dataset.

```python
def formata_cnpj(d: str) -> str:
    return f"{d[:2]}.{d[2:5]}.{d[5:8]}/{d[8:12]}-{d[12:]}"

def normaliza_cnpj(cnpj: str) -> str | None:
    d = so_digitos(cnpj)
    if len(d) != 14:
        return None
    return formata_cnpj(d)

df["cnpj_normalizado"] = df["cnpj"].apply(normaliza_cnpj)
```

Definimos a função que vai limpar, validar e formatar números de telefone.

Definimos d para remover tudo que não for número (parênteses, espaços, traços, etc.).

Fazemos um If que se a string estiver vazia, retorna None.

Logo, fazemos a remoção do código do país (+55). Caso o número comece com o código internacional do Brasil, e, tenha mais de 11 dígitos, ele é removido.

11 dígitos → já tem o “9” inicial dos celulares.

10 dígitos → número fixo antigo (sem o 9).

Usamos um return que eixa no formato padrão brasileiro: (DDD) 9XXXX-XXXX

Aplicamos a função à coluna "telefone" e salva o resultado em "telefone_normalizado".

```python
def normaliza_telefone(fone: str) -> str | None:
    d = so_digitos(fone)
    if not d:
        return None

    # Remove código do país 55 se vier
    if d.startswith("55") and len(d) > 11:
        d = d[2:]

    # Aceita 10 ou 11 dígitos (DDD + número)
    if len(d) == 11:
        ddd = d[:2]
        num = d[2:]
    elif len(d) == 10:
        ddd = d[:2]
        num = "9" + d[2:]  # insere o 9 no começo
    else:
        return None  # demais casos eu marco como inválido

    return f"({ddd}) {num[0]}{num[1:5]}-{num[5:]}"

df["telefone_normalizado"] = df["telefone"].apply(normaliza_telefone)
```

Definimos a função que irá normalizar os CEPs, onde irá receber um texto representando um CEP e retorna o CEP normalizado ou None se estiver inválido.

Chamamos a função auxiliar so_digitos que remove todos os caracteres não numéricos, mantendo só os dígitos.


Fazemos um len que verifica se o CEP tem exatamente 8 dígitos, o formato válido no Brasil. Caso não tenha 8 dígitos, retorna None (valor nulo indicando CEP inválido).

Se estiver correto, formata o CEP como #####-###.

Aplica a função em todas as linhas da coluna cep do dataset, criando uma nova coluna com o CEP padronizado.

```python
def normaliza_cep(cep: str) -> str | None:
    d = so_digitos(cep)
    if len(d) != 8:
        return None
    return f"{d[:5]}-{d[5:]}"

df["cep_normalizado"] = df["cep"].apply(normaliza_cep)
```

Fazemos uma função que irá normalizar as datas, onde verifica se o valor é uma string; caso contrário, retorna None.

Usa uma regex para capturar uma data no formato dd/mm/aaaa, mesmo que tenha outros separadores (-, ., espaço, etc.). Se nada for encontrado, retorna None.

Converte os valores para inteiros "(map(int, m.groups()))". E validamos o ano (de 1900 a 2100). Criamos um objeto datetime.date. Se for uma data inválida (como 31/02/2023), dá erro e retorna None. Se tudo estiver certo, retorna a data formatada como "dd/mm/yyyy".

Cria a coluna data_normalizada que será adicionada no dataset.

```python
def normaliza_data(texto: str) -> str | None:
    if not isinstance(texto, str):
        return None

    # pega a primeira sequência dd qualquer / mm qualquer / aaaa (4 dígitos)
    m = re.search(r"(\d{1,2})\D+(\d{1,2})\D+(\d{4})", texto)
    if not m:
        return None

    d, mth, y = map(int, m.groups())

    # restringe ano para algo razoável
    if not (1900 <= y <= 2100):
        return None

    try:
        dt = datetime.date(y, mth, d)
    except ValueError:
        return None

    return dt.strftime("%d/%m/%Y")

df["data_normalizada"] = df["data"].apply(normaliza_data)
```

Fazemos uma função para normalizar o valor monetário, que irá garantir que o valor é uma string.

Primeira busca: formato brasileiro de valores (ex: 1.234,56), com possíveis pontos de milhar e vírgula decimal. Remove pontos, troca vírgula por ponto e converte em float. Se não achar, tenta formato internacional (1234.56). Se ainda não achar, tenta extrair um número inteiro simples. Retorna None se nada for encontrado.

Cria a coluna valor_float com os valores numéricos extraídos.

```python
def extrai_valor_monetario(texto: str) -> float | None:
    if not isinstance(texto, str):
        return None

    # padrão brasileiro com milhar e vírgula: 1.234,56
    m = re.search(r"-?\d{1,3}(?:\.\d{3})*(?:,\d{2})", texto)
    if m:
        s = m.group(0).replace(".", "").replace(",", ".")
        try:
            return float(s)
        except ValueError:
            return None

    # fallback: 1234.56
    m = re.search(r"-?\d+\.\d{2}", texto)
    if m:
        try:
            return float(m.group(0))
        except ValueError:
            return None

    # último recurso: inteiro
    m = re.search(r"-?\d+", texto)
    return float(m.group(0)) if m else None

df["valor_float"] = df["valor"].apply(extrai_valor_monetario)
```

Nas funções para normalizar os emails e urls, irão fazer quase a mesma coisa. Cada uma delas garantem que o texto é string.

Removem espaços indevidos. Procuram o primeiro e-mail ou URL válido no texto Retornam None se não encontrar.

Cria duas novas colunas com os valores limpos e padronizados.

```python
def normaliza_email(texto: str) -> str | None:
    if not isinstance(texto, str):
        return None
    # remove espaços no meio, que quebram o e-mail
    m = RX_EMAIL.search(texto.replace(" ", ""))
    return m.group(0) if m else None

def normaliza_url(texto: str) -> str | None:
    if not isinstance(texto, str):
        return None
    m = RX_URL.search(texto.replace(" ", ""))
    return m.group(0) if m else None

df["email_normalizado"] = df["email"].apply(normaliza_email)
df["url_normalizada"]   = df["url"].apply(normaliza_url)
```

**Etapa 4 — VALIDAÇÃO**

Para começarmos as validações, fazemos primeiro a validação dos CPFs, usando a função so_digitos, que irá remove tudo que não for número, deixando apenas os 11 dígitos do CPF. Exemplo: "123.456.789-09" → "12345678909"


Fazemos um len que verifica que cada CPF válido sempre tem 11 dígitos numéricos.
Se tiver a mais ou a menos, o CPF já é considerado inválido.

Fazemos outro len para que cada linha detecta CPFs compostos por números repetidos, tipo: 11111111111 ou 00000000000.

set(d) transforma a string em um conjunto (sem repetições).
Se o tamanho do conjunto for 1, significa que todos os dígitos são iguais → inválido.

O CPF tem 9 dígitos base e 2 dígitos verificadores no final.
Aqui estamos calculando o primeiro deles. Logo, multiplica cada um dos 9 primeiros dígitos por um peso decrescente de 10 a 2. Soma todos esses produtos. Calcula o resto da divisão por 11 (resto = soma % 11), se resto < 2, o primeiro dígito verificador (dig1) é 0; caso contrário, é 11 - resto.

Comparamos o dígito calculado com o dígito real do CPF (posição 9, já que começa em 0). Se não bater, o CPF é inválido. Fazemos a soma para que o peso vá de 11 a 2, pois incluímos também o primeiro dígito verificador.

Fazemos a verificação final, ode compara o segundo dígito calculado com o segundo dígito original do CPF. Se tudo passar, retorna True.

Aplica a função a cada valor da coluna cpf, criando uma coluna booleana (True = válido, False = inválido).

```python
def valida_cpf(cpf: str) -> bool:
    d = so_digitos(cpf)
    if len(d) != 11:
        return False

    # descarta sequências tipo 11111111111
    if len(set(d)) == 1:
        return False

    # primeiro dígito verificador
    soma = sum(int(d[i]) * (10 - i) for i in range(9))
    resto = soma % 11
    dig1 = 0 if resto < 2 else 11 - resto
    if dig1 != int(d[9]):
        return False

    # segundo dígito verificador
    soma = sum(int(d[i]) * (11 - i) for i in range(10))
    resto = soma % 11
    dig2 = 0 if resto < 2 else 11 - resto
    if dig2 != int(d[10]):
        return False

    return True

df["cpf_valido"] = df["cpf"].apply(valida_cpf)
```

Fazemos a função que valida os CNPJs, onde usamos a função "so_digitos" que remove tudo que não é número. Exemplo: "12.345.678/0001-95" → "12345678000195"

Fazemos um len que verifica se CNPJs válidos sempre têm 14 dígitos.

Fazemos outro len que elimina casos como "00000000000000".

Para o primeiro dígito verificador, cada número é multiplicado por um peso específico. O peso começa em 5 e vai até 2, depois recomeça em 9 e vai até 2 novamente. Soma-se tudo, tira o módulo 11: Se resto < 2, o dígito é 0. Senão, é 11 - resto.

Fazemos a verificação do primeiro digito, o dígito calculado é comparado com o 13º caractere do CNPJ (posição 12). Logo, o calculo do segundo digito verificador, o processo é igual, mas adiciona o peso 6 no início, totalizando 13 pesos.

Fazemos a verificação final, que compara o segundo dígito calculado com o último número do CNPJ. Se tudo for válido = True.

Por ultimo, criamos a coluna cnpj_valido com valores booleanos, que será incluida no dataset.

```python
def valida_cnpj(cnpj: str) -> bool:
    d = so_digitos(cnpj)
    if len(d) != 14:
        return False

    if len(set(d)) == 1:
        return False

    pesos1 = [5,4,3,2,9,8,7,6,5,4,3,2]
    soma = sum(int(d[i]) * pesos1[i] for i in range(12))
    resto = soma % 11
    dig1 = 0 if resto < 2 else 11 - resto
    if dig1 != int(d[12]):
        return False

    pesos2 = [6] + pesos1
    soma = sum(int(d[i]) * pesos2[i] for i in range(13))
    resto = soma % 11
    dig2 = 0 if resto < 2 else 11 - resto
    if dig2 != int(d[13]):
        return False

    return True

df["cnpj_valido"] = df["cnpj"].apply(valida_cnpj)
```

**Etapa 5 — RELATÓRIO**

Nesta etapa, faremos o relatório dos valores obtidos com as normalizações usadas nas colunas que precisavam no nosso dataset.

Primeiro, fazemos um n_total, que conta quantas linhas há no dataframe (número total de registros).

Logo, criamos um dicionário, que cada chave representa uma métrica. Cada valor é calculado com expressões baseadas nas colunas normalizadas e validadas.

Exemplos:
df["cpf_valido"].sum()
Soma os valores True (Python interpreta True como 1 e False como 0), dando o número de CPFs válidos.

(~df["cpf_valido"]).sum()
O operador ~ inverte os valores booleanos (True vira False e vice-versa).
Assim, soma os inválidos.

df["telefone_normalizado"].notna().sum()
Conta quantos registros têm telefone não nulo (ou seja, foi normalizado com sucesso).

Isso é repetido para CEPs, datas, e-mails, URLs e valores monetários. Exibimos o dicionário de métricas e o total de linhas, permitindo calcular porcentagens depois. Onde obtemos os resultados abaixo:

```python
n_total = len(df)

resumo = {
    "cpf_validos": int(df["cpf_valido"].sum()),
    "cpf_invalidos": int((~df["cpf_valido"]).sum()),
    "cnpj_validos": int(df["cnpj_valido"].sum()),
    "cnpj_invalidos": int((~df["cnpj_valido"]).sum()),
    "nomes_normalizados": int(df["nome_normalizado"].notna().sum()),
    "cpfs_normalizados": int(df["cpf_normalizado"].notna().sum()),
    "telefones_normalizados": int(df["telefone_normalizado"].notna().sum()),
    "ceps_normalizados": int(df["cep_normalizado"].notna().sum()),
    "datas_normalizadas": int(df["data_normalizada"].notna().sum()),
    "emails_normalizados": int(df["email_normalizado"].notna().sum()),
    "urls_normalizadas": int(df["url_normalizada"].notna().sum()),
    "valores_preenchidos": int(df["valor_float"].notna().sum()),
}

resumo, n_total
```
