```python
import pandas as pd
import re
import numpy as np


df = pd.read_csv('dataset_sintetico_regex_ruidos.csv')
# ============================================
# 🧠 Análise de Dados Textuais com Regex
# Autor: Jefté Damaceno de Moura
# ============================================



# ============================
# ETAPA 1 — CARREGAMENTO E EXPLORAÇÃO
# ============================


# Visualizar as primeiras linhas
print("📊 Visualização inicial do dataset:")
display(df.head())

# Ver informações básicas
print("\n📋 Informações gerais:")
print(df.info())

# ============================
# ETAPA 2 — CRIAÇÃO DE PADRÕES REGEX
# ============================

padroes = {
    "nome": r"^[A-ZÁÉÍÓÚÂÊÔÃÕÇ][a-záéíóúâêôãõç]+(\s[A-ZÁÉÍÓÚÂÊÔÃÕÇ][a-záéíóúâêôãõç]+)+$",
    "cpf": r"^\d{3}\.?\d{3}\.?\d{3}-?\d{2}$",
    "cnpj": r"^\d{2}\.?\d{3}\.?\d{3}/?\d{4}-?\d{2}$",
    "email": r"^[\w\.-]+@[\w\.-]+\.\w+$",
    "telefone": r"^\(?\d{2}\)?\s?\d{4,5}-?\d{4}$",
    "cep": r"^\d{5}-?\d{3}$",
    "data": r"^(0[1-9]|[12][0-9]|3[01])/(0[1-9]|1[0-2])/[0-9]{4}$",
    "valor": r"^R\$ ?\d{1,3}(\.\d{3})*,\d{2}$"
}

# ============================
#  TESTE DOS PADRÕES
# ============================

def testar_regex(coluna, padrao):
    """Retorna se o valor está válido (True) ou não (False) com base no regex."""
    return df[coluna].apply(lambda x: bool(re.fullmatch(padrao, str(x).strip())))

# Aplica os padrões a todas as colunas que existirem no dataset
resultados = {}
for campo, padrao in padroes.items():
    if campo in df.columns:
        resultados[campo] = testar_regex(campo, padrao)
        print(f"\n🔍 Resultados para '{campo}':")
        print(resultados[campo].value_counts())

# ============================
#  RELATÓRIO DE QUALIDADE
# ============================

print("\n📈 RELATÓRIO DE CAMPOS INVÁLIDOS:")
for campo, validacoes in resultados.items():
    invalidos = df[~validacoes]
    if len(invalidos) > 0:
        print(f"\n❌ {campo.upper()} - {len(invalidos)} inválidos")
        display(invalidos[[campo]].head())
    else:
        print(f"\n✅ {campo.upper()} - Todos os valores estão válidos!")

```

```python
# ============================================
# 🧠 Etapas 3 e 4 — Normalização e Validação
# Autor: Jefté Damaceno de Moura
# ============================================

import pandas as pd
import re
from datetime import datetime
import unicodedata
import numpy as np

# ============================
# ETAPA 0 — CONFIGURAÇÃO INICIAL
# ============================

NOMES_COMUNS = {
    "ana", "maria", "joao", "luiz", "pedro", "paula", "fernanda", "carlos", "lucas",
    "juliana", "bruno", "marcos", "camila", "rafael", "beatriz", "gabriel", "larissa",
    "roberto", "rodrigo", "aline", "ricardo", "andre", "felipe", "carla"
}

SOBRENOMES_COMUNS = {
    "silva", "santos", "souza", "oliveira", "pereira", "lima", "rodrigues",
    "alves", "gomes", "ribeiro", "costa", "carvalho", "araujo", "melo",
    "barbosa", "castro", "martins", "dias", "teixeira", "freitas", "tavares",
    "machado", "moura", "nogueira", "fonseca", "pires", "campos", "coelho",
    "moreira", "almeida", "jesus", "cardoso"
}

def separar_nome_colado(nome: str) -> str:
    if not isinstance(nome, str) or not nome.strip():
        return nome
    nome = nome.lower().strip()
    for sobrenome in sorted(SOBRENOMES_COMUNS, key=len, reverse=True):
        if nome.endswith(sobrenome):
            prefixo = nome[:-len(sobrenome)]
            for n in sorted(NOMES_COMUNS, key=len, reverse=True):
                if prefixo.startswith(n):
                    return f"{n.capitalize()} {sobrenome.capitalize()}"
    for n1 in sorted(NOMES_COMUNS, key=len, reverse=True):
        if nome.startswith(n1):
            resto = nome[len(n1):]
            for n2 in sorted(NOMES_COMUNS, key=len, reverse=True):
                if resto.startswith(n2):
                    return f"{n1.capitalize()} {n2.capitalize()}"
    return nome.capitalize()

# ============================
# ETAPA 1 — LEITURA DO DATASET
# ============================

df = pd.read_csv('dataset_sintetico_regex_ruidos.csv')

print("📊 Dados originais:")
display(df.head())

# ============================
# ETAPA 3 — NORMALIZAÇÃO
# ============================

def limpar_numericos(campo):
    if pd.isna(campo):
        return ""
    return re.sub(r'[^0-9]', '', str(campo))

def limpar_nome(nome):
    if pd.isna(nome):
        return nome
    nome = re.sub(r'(\b\w)\s+(\w\b)', r'\1\2', nome)
    nome = re.sub(r'([a-z])([A-ZÁÉÍÓÚÂÊÔÃÕÇ])', r'\1 \2', nome)
    nome = nome.strip()
    nome = separar_nome_colado(nome)
    nome = ' '.join([n.capitalize() for n in nome.split()])
    return nome.strip()

# ============================
# CPF / CNPJ / TELEFONE
# ============================

def normaliza_cpf(cpf):
    """Formata CPF para XXX.XXX.XXX-YY sem alterar os dígitos. Se inválido → NaN"""
    if pd.isna(cpf):
        return np.nan
    cpf_limpo = limpar_numericos(cpf)
    if len(cpf_limpo) == 11:
        return f"{cpf_limpo[:3]}.{cpf_limpo[3:6]}.{cpf_limpo[6:9]}-{cpf_limpo[9:]}"
    else:
        return np.nan

# def normaliza_cnpj(cnpj):
#     """Formata CNPJ para XX.XXX.XXX/0001-YY sem alterar dígitos. Se inválido → NaN"""
#     if pd.isna(cnpj):
#         return np.nan
#     cnpj_limpo = limpar_numericos(cnpj)
#     if len(cnpj_limpo) == 14:
#         # Mantém a raiz e sufixo originais
#         return f"{cnpj_limpo[:2]}.{cnpj_limpo[2:5]}.{cnpj_limpo[5:8]}/{cnpj_limpo[8:12]}-{cnpj_limpo[12:]}"
#     else:
#         return np.nan
def normaliza_cnpj(cnpj):
    """Padroniza CNPJ mesmo com pequenos ruídos — mantém dígitos originais, se possível."""
    if pd.isna(cnpj):
        return np.nan
    cnpj_limpo = limpar_numericos(cnpj)
    if len(cnpj_limpo) < 13 or len(cnpj_limpo) > 15:
        return np.nan  # rejeita se for absurdo
    # Corrige se tiver 13 ou 15 dígitos (provável erro de digitação)
    if len(cnpj_limpo) == 13:
        cnpj_limpo = "0" + cnpj_limpo
    elif len(cnpj_limpo) == 15:
        cnpj_limpo = cnpj_limpo[:14]
    # Formata padrão
    try:
        return f"{cnpj_limpo[:2]}.{cnpj_limpo[2:5]}.{cnpj_limpo[5:8]}/{cnpj_limpo[8:12]}-{cnpj_limpo[12:]}"
    except:
        return np.nan

def normaliza_fone(fone):
    """Padroniza telefone para (XX) 9XXXX-XXXX"""
    if pd.isna(fone):
        return fone
    fone = limpar_numericos(fone)
    if len(fone) == 11:
        return f"({fone[:2]}) {fone[2:3]}{fone[3:7]}-{fone[7:]}"
    elif len(fone) == 10:
        return f"({fone[:2]}) {fone[2:6]}-{fone[6:]}"
    else:
        return np.nan

# ============================
# DATAS — corrigir e padronizar
# ============================

def normaliza_data(data):
    """Corrige e padroniza datas comuns no formato DD-MM-YYYY."""
    if pd.isna(data):
        return np.nan
    data_str = str(data).strip()
    data_str = re.sub(r'[^0-9/.-]', '', data_str)  # remove letras
    data_str = re.sub(r'//+', '/', data_str)        # corrige barras duplas
    data_str = re.sub(r'\s+', '', data_str)         # remove espaços
    data_str = data_str.replace('.', '/')           # troca pontos por barras

    # tenta adicionar zeros faltantes em dia/mês
    partes = re.split(r'[-/]', data_str)
    if len(partes) == 3:
        dia, mes, ano = partes
        if len(dia) == 1: dia = dia.zfill(2)
        if len(mes) == 1: mes = mes.zfill(2)
        if len(ano) == 2: ano = "20" + ano  # ex: 05 -> 2005
        data_str = f"{dia}/{mes}/{ano}"

    formatos = ["%d/%m/%Y", "%d-%m-%Y", "%Y-%m-%d"]
    for fmt in formatos:
        try:
            dt = datetime.strptime(data_str, fmt)
            return dt.strftime("%d-%m-%Y")
        except:
            pass
    return np.nan


# ============================
# OUTRAS FUNÇÕES
# ============================

def normaliza_email(email):
    if pd.isna(email):
        return np.nan
    email = str(email).strip().lower()
    email = re.sub(r'\s+', '', email)
    return email if "@" in email else np.nan

def normaliza_url(url):
    if pd.isna(url):
        return np.nan
    url = str(url).strip().lower()
    if not url.startswith("http"):
        url = "https://" + url
    return url

def normaliza_valor(valor):
    if pd.isna(valor):
        return np.nan
    valor_str = str(valor)
    valor_str = re.sub(r'[^\d,.-]', '', valor_str)
    valor_str = valor_str.replace('.', '').replace(',', '.')
    try:
        return round(float(valor_str), 2)
    except:
        return np.nan

def normaliza_texto_livre(txt):
    if pd.isna(txt):
        return np.nan
    txt = str(txt).strip()
    txt = unicodedata.normalize('NFKD', txt)
    txt = re.sub(r'[^A-Za-z0-9\s.,;!?-]', '', txt)
    txt = re.sub(r'\s+', ' ', txt)
    return txt.strip().capitalize()

# ============================
# APLICAÇÃO
# ============================

if 'nome' in df.columns:
    df['nome'] = df['nome'].apply(limpar_nome)
if 'cpf' in df.columns:
    df['cpf_normalizado'] = df['cpf'].apply(normaliza_cpf)
if 'cnpj' in df.columns:
    df['cnpj_normalizado'] = df['cnpj'].apply(normaliza_cnpj)
if 'telefone' in df.columns:
    df['telefone_normalizado'] = df['telefone'].apply(normaliza_fone)
if 'data' in df.columns:
    df['data_normalizada'] = df['data'].apply(normaliza_data)
if 'email' in df.columns:
    df['email_normalizado'] = df['email'].apply(normaliza_email)
if 'url' in df.columns:
    df['url_normalizada'] = df['url'].apply(normaliza_url)
if 'valor' in df.columns:
    df['valor_normalizado'] = df['valor'].apply(normaliza_valor)
if 'texto_livre' in df.columns:
    df['texto_livre_normalizado'] = df['texto_livre'].apply(normaliza_texto_livre)

print("✅ Dados após normalização completa:")
display(df.head())

# ============================
# ETAPA 4 — VALIDAÇÃO
# ============================

def validar_cpf(cpf):
    cpf = limpar_numericos(cpf)
    if len(cpf) != 11 or cpf == cpf[0] * 11:
        return False
    soma1 = sum(int(cpf[i]) * (10 - i) for i in range(9))
    d1 = (soma1 * 10 % 11) % 10
    soma2 = sum(int(cpf[i]) * (11 - i) for i in range(10))
    d2 = (soma2 * 10 % 11) % 10
    return d1 == int(cpf[9]) and d2 == int(cpf[10])

def validar_cnpj(cnpj):
    cnpj = limpar_numericos(cnpj)
    if len(cnpj) != 14:
        return False
    pesos1 = [5,4,3,2,9,8,7,6,5,4,3,2]
    pesos2 = [6] + pesos1
    soma1 = sum(int(cnpj[i]) * pesos1[i] for i in range(12))
    d1 = 11 - soma1 % 11
    if d1 >= 10: d1 = 0
    soma2 = sum(int(cnpj[i]) * pesos2[i] for i in range(13))
    d2 = 11 - soma2 % 11
    if d2 >= 10: d2 = 0
    return d1 == int(cnpj[12]) and d2 == int(cnpj[13])

if 'cpf_normalizado' in df.columns:
    df.loc[~df['cpf_normalizado'].apply(lambda x: validar_cpf(x) if pd.notna(x) else False), 'cpf_normalizado'] = np.nan
if 'cnpj_normalizado' in df.columns:
    df.loc[~df['cnpj_normalizado'].apply(lambda x: validar_cnpj(x) if pd.notna(x) else False), 'cnpj_normalizado'] = np.nan

print("📋 Validação de CPFs e CNPJs finalizada:")
display(df[['cpf_normalizado', 'cnpj_normalizado']].head())

# ============================
# ETAPA 5 — EXPORTAÇÃO
# ============================

df.to_csv('dados_ajustados.csv', index=False)
print("\n💾 Arquivo salvo como 'dados_ajustados.csv'")
```

```python
# =======================================================
# 🧠 Comparação entre Datasets — Antes x Depois da Limpeza
# =======================================================

import pandas as pd
import re
import numpy as np

# ============================
# ETAPA 0 — CARREGAMENTO
# ============================

# Dataset original (antes da limpeza)
df_original = pd.read_csv('dataset_sintetico_regex_ruidos.csv')

# Dataset ajustado (após normalização e validação)
df_limpo = pd.read_csv('dados_ajustados.csv')

print("📊 Visualização dos dados originais:")
display(df_original.head())

print("📊 Visualização dos dados ajustados:")
display(df_limpo.head())

# ============================
# ETAPA 1 — MAPEAMENTO DE COLUNAS
# ============================

# Define pares de comparação (coluna original → coluna ajustada)
mapeamento_colunas = {
    "cpf": "cpf_normalizado",
    "cnpj": "cnpj_normalizado",
    "telefone": "telefone_normalizado",
    "email": "email_normalizado",
    "url": "url_normalizada",
    "valor": "valor_normalizado",
    "data": "data_normalizada",
    "texto_livre": "texto_livre_normalizado",
    "nome": "nome"  # mesmo nome em ambos
}

# Filtra apenas colunas existentes em ambos
pares_validos = {
    orig: novo for orig, novo in mapeamento_colunas.items()
    if orig in df_original.columns and novo in df_limpo.columns
}

print(f"\n🔗 Colunas comparadas: {pares_validos}\n")

# ============================
# ETAPA 2 — COMPARAÇÃO DIRETA
# ============================

comparacao = pd.DataFrame()

for orig, novo in pares_validos.items():
    comparacao[f"{orig}_original"] = df_original[orig].astype(str).str.strip()
    comparacao[f"{orig}_ajustado"] = df_limpo[novo].astype(str).str.strip()
    comparacao[f"{orig}_igual"] = comparacao[f"{orig}_original"].eq(comparacao[f"{orig}_ajustado"])

# Mostra apenas registros com diferenças
diferencas = comparacao.loc[~comparacao.filter(like='_igual').all(axis=1)]

print("\n📋 DIFERENÇAS ENTRE OS DATASETS:")
display(diferencas.head(20))

# ============================
# ETAPA 3 — RESUMO DAS DIFERENÇAS
# ============================

print("\n📈 RESUMO DAS DIFERENÇAS:")
for orig, novo in pares_validos.items():
    iguais = comparacao[f"{orig}_igual"].sum()
    total = len(comparacao)
    diferentes = total - iguais
    pct = (diferentes / total) * 100
    print(f"🧩 {orig}: {diferentes}/{total} ({pct:.1f}%) valores diferentes")

# ============================
# ETAPA 4 — ANÁLISE DE QUALIDADE (REGEX)
# ============================

# Padrões atualizados conforme a nova normalização
padroes = {
    "nome": r"^[A-ZÁÉÍÓÚÂÊÔÃÕÇ][a-záéíóúâêôãõç]+(\s[A-ZÁÉÍÓÚÂÊÔÃÕÇ][a-záéíóúâêôãõç]+)+$",
    "cpf_normalizado": r"^\d{3}\.\d{3}\.\d{3}-\d{2}$",
    "cnpj_normalizado": r"^\d{2}\.\d{3}\.\d{3}/\d{4}-\d{2}$",
    "email_normalizado": r"^[\w\.-]+@[\w\.-]+\.\w+$",
    "telefone_normalizado": r"^\(\d{2}\)\s9\d{4}-\d{4}$",
    "cep": r"^\d{5}-\d{3}$",
    "data_normalizada": r"^\d{2}-\d{2}-\d{4}$",
    "valor_normalizado": r"^\d+(\.\d{2})?$"
}

def testar_regex(df, coluna, padrao):
    return df[coluna].apply(lambda x: bool(re.fullmatch(padrao, str(x).strip())) if pd.notna(x) else False)

print("\n📊 RELATÓRIO DE VALIDAÇÃO (ANTES x DEPOIS):")
for orig, novo in pares_validos.items():
    padrao = padroes.get(novo, None)
    if padrao:
        valid_orig = testar_regex(df_original, orig, padrao)
        valid_novo = testar_regex(df_limpo, novo, padrao)
        print(f"\n🔎 {orig.upper()}:")
        print(f"Antes: {valid_orig.sum()} válidos / {len(valid_orig)} ({valid_orig.mean()*100:.1f}%)")
        print(f"Depois: {valid_novo.sum()} válidos / {len(valid_novo)} ({valid_novo.mean()*100:.1f}%)")

# ============================
# ETAPA 5 — EXPORTAÇÃO OPCIONAL
# ============================

diferencas.to_csv('comparacao_datasets.csv', index=False)
print("\n💾 Arquivo de diferenças salvo como 'comparacao_datasets.csv'")
```
